**Trier des fichiers**

```python
import os
import shutil
from pathlib import Path

# --- Configuration des Chemins (CORRIGÉE) ---
# Définit le chemin de base du script en utilisant le répertoire de travail actuel (compatible Jupyter)
BASE_DIR = Path.cwd()

# Le dossier où se trouvent les fichiers en vrac
DOSSIER_SOURCE = BASE_DIR / "fichiers_en_vrac"

# --- Configuration des Extensions et des Dossiers de Destination ---
EXTENSIONS_FICHIERS = {
    "Images": [".jpg", ".jpeg", ".png", ".gif", ".tiff"],
    "Documents": [".doc", ".docx", ".odt", ".xls", ".xlsx", ".pptx"],
    "Audio": [".mp3", ".wav", ".flac", ".aac"],
    "Vidéos": [".mp4", ".avi", ".mov", ".mkv"],
    "PDF": [".pdf"],
    "Archives": [".zip", ".rar", ".7z"],
    "Scripts": [".py", ".sh", ".js", ".html", ".css"]
}

# --- Fonction de Tri ---

def trier_fichiers():
    """Trie les fichiers du DOSSIER_SOURCE dans des sous-dossiers basés sur l'extension."""
    
    # 1. Vérification et initialisation
    if not DOSSIER_SOURCE.is_dir():
        print(f"ERREUR : Le dossier source '{DOSSIER_SOURCE}' n'existe pas ou n'est pas un répertoire.")
        return 

    print(f"\n--- Démarrage du tri dans le dossier : {DOSSIER_SOURCE} ---")
    fichiers_tries = 0
    fichiers_inconnus = 0
    
    # 2. Créer les dossiers de destination si nécessaire
    dossiers_dest = {}
    for nom_dossier in EXTENSIONS_FICHIERS.keys():
        chemin_dest = DOSSIER_SOURCE / nom_dossier
        chemin_dest.mkdir(exist_ok=True)
        dossiers_dest[nom_dossier] = chemin_dest
    
    # 3. Parcourir et trier les fichiers
    for item in DOSSIER_SOURCE.iterdir():
        if item.is_file():
            extension = item.suffix.lower()
            fichier_trie = False
            
            # Recherche de la catégorie correspondante
            for categorie, extensions_list in EXTENSIONS_FICHIERS.items():
                if extension in extensions_list:
                    chemin_destination = dossiers_dest[categorie] / item.name
                    
                    try:
                        shutil.move(str(item), str(chemin_destination))
                        print(f"Déplacé {item.name} vers /{categorie}/")
                        fichiers_tries += 1
                        fichier_trie = True
                        break
                    except Exception as e:
                        print(f"ERREUR lors du déplacement de {item.name} : {e}")
                        fichier_trie = True
                        break
            
            if not fichier_trie:
                print(f"ATTENTION : Fichier inconnu ({extension}) : {item.name}")
                fichiers_inconnus += 1
                
    print(f"\n--- Fin du tri ---")
    print(f"Statistiques : {fichiers_tries} fichiers triés, {fichiers_inconnus} fichiers inconnus.")


if __name__ == "__main__":
    
    # Crée le dossier source s'il n'existe pas (pour les tests)
    DOSSIER_SOURCE.mkdir(parents=True, exist_ok=True)
    
    # 1. Exécuter la fonction de tri
    trier_fichiers()
```
