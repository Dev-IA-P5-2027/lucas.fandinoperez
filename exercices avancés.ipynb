```python
# Vous devez créer une fonction qui trie automatiquement des fichiers dans des sous-dossiers en fonction de leur type (et donc de leur extension). Facultatif : Réaliser la fonction inverse → déplacer les fichiers des sous-dossiers dans le dossier principal.

import json
import os
import shutil

DOSSIER_SOURCE = "C:/Users/lucas.fandinoperez/Desktop/github/lucas.fandinoperez/fichiers en vrac"

EXTENSIONS_FICHIERS = {"Images": [".jpg", ".jpeg", ".png"], "Documents": [".doc",".docx", ".odt", ".xls", ".xlsx" ], "Audio": [".mp3", ".wav", ".flac", ".aac"], "Vidéos": [".mp4", ".avi", ".mov", ".mkv"], "PDF": [".pdf"]}

DOSSIER_DOCS = "C:/Users/lucas.fandinoperez/Desktop/github/lucas.fandinoperez/Documents"
DOSSIER_IMGS = "C:/Users/lucas.fandinoperez/Desktop/github/lucas.fandinoperez/Images"
DOSSIER_MUSIQUE = "C:/Users/lucas.fandinoperez/Desktop/github/lucas.fandinoperez/Musiques"
DOSSIER_VIDEO = "C:/Users/lucas.fandinoperez/Desktop/github/lucas.fandinoperez/Videos"



def trier_fichiers(filepath):
    if not os.path.isdir(DOSSIER_SOURCE):
        print(f"ERREUR : {DOSSIER_SOURCE} n'est pas un dossier valide.")
        return 
        
    
    for fichier in os.listdir(DOSSIER_SOURCE):
        chemin_source_complet = os.path.join(DOSSIER_SOURCE, fichier)
        if os.path.isdir(chemin_source_complet):
            continue
        extension = os.path.splitext(fichier)
        extension = extension[1].lower()
        for k, v in EXTENSIONS_FICHIERS.items():
            if extension in v:
                categorie_trouvee = k
                if categorie_trouvee:
                    if extension in EXTENSIONS_FICHIERS["Documents"] and EXTENSIONS_FICHIERS["PDF"]:
                        shutil.move(chemin_source_complet, DOSSIER_DOCS)
                        print(f"Le fichier {fichier} a bien été déplacé de {DOSSIER_SOURCE} vers {DOSSIER_DOCS}.")
                    elif extension in EXTENSIONS_FICHIERS["Images"]:
                        shutil.move(chemin_source_complet, DOSSIER_IMGS)
                        print(f"Le fichier {fichier} a bien été déplacé de {DOSSIER_SOURCE} vers {DOSSIER_IMGS}.")
                    elif extension in EXTENSIONS_FICHIERS["Audio"]:
                        shutil.move(chemin_source_complet, DOSSIER_MUSIQUE)
                        print(f"Le fichier {fichier} a bien été déplacé de {DOSSIER_SOURCE} vers {DOSSIER_MUSIQUE}.")
                    elif extension in EXTENSIONS_FICHIERS["Vidéos"]:
                        shutil.move(chemin_source_complet, DOSSIER_VIDEO)
                        print(f"Le fichier {fichier} a bien été déplacé de {DOSSIER_SOURCE} vers {DOSSIER_VIDEO}.")
            else:
                print(f"ATTENTION ! Le fichier {fichier} est de format inconnu ({extension}) , le fichier ne peut pas être trié.")
                break
    print("Tous les fichiers ont été déplacés avec succès, fin du programme.")






trier_fichiers(DOSSIER_SOURCE)
        
    


```
