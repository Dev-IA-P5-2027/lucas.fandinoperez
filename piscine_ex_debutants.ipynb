```python
# Ecrire un programme en langage Python qui demande à l'utilisateur de saisir deux nombres entiers a et b et de lui afficher le quotient et le reste de la division euclidienne de a par b.
number_a = int(input("Veuillez entrer un premier nombre a :"))
number_b = int(input("Veuillez entrer un deuxième nombre b :"))


if number_a > number_b:
    quotient = number_a // number_b
    reste = number_a % number_b
    print(f"Le quotient est {quotient} et le reste est {reste}")
elif number_a < number_b:
    quotient = number_b // number_a
    reste = number_b % number_a
    print(f"Le quotient est {quotient} et le reste est {reste}")
else:
    print("Les nombres a et b sont egaux, veuillez choisir deux nombres differents.")
```

```python
# Ecrire un programme en langage Python, permettant d’échanger le premier et le dernier caractère d’une chaine donnée.

s= str(input("Veuillez saisir un mot svp: "))
liste = []

for letter in s:
    liste.append(letter)
    reversed_list = liste[::-1]
print(f"Mot de base {s}. Après inversion, le premier caractère est {reversed_list[0]} et le dernier est {reversed_list[-1]}.")
```

```python
# Un palindrome est un mot dont l'ordre des lettres reste le même si on le lit de gauche à droite ou de droite à gauche. Par exemple : 'laval' , 'radar, 'sos'... sont des palindromes. Ecrire un programme en Python qui demande à l'utilisateur de saisir un mot et de lui renvoyer s'il s'agit d'un palindrome ou non?
s= str(input("Veuillez saisir un mot svp: "))
mot_inverse = s[::-1]

if s == mot_inverse:
    print("Le mot donné est un palindrome.")
else:
    print("Le mot donné n'est pas un palindrome.")
```
