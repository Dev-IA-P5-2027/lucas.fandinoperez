```python
# Utilisez une exception pour calculer, dans une boucle évoluant de -3 à 3 compris, la valeur de sin(x)/x. L'exception permettra de gérer l'erreur rencontrée, de l'afficher et de continuer le programme.
import math

def calc_sinx():
    for x in range(-3, 4, 1):
        try:
            resultat = math.sin(x)/x
            print(f"sin({x}) divisé par {x} est égal à {resultat}.")
        except ZeroDivisionError:
            print(f"Impossible de diviser par zéro pour l'élément {x}")

calc_sinx()
```

```python
# Écrire une fonction qui retourne f(x) = 2x3 + x − 5. Écrire une fonction poly avec quatre paramètres : fonction, borneInf, borneSup et nbPas. Cette procédure affiche les valeurs de fonction, de borneInf à borneSup, touslesnbPas. Elle doit respecter borne Inf < borne Sup.

# La fonction poly doit renvoyer un dictionnaire clé:valeur (clé = x, valeur = f(x)).

def calc_x(x):
    return 2*x**3 + x - 5

def poly(calc_x, borneInf, borneSup, nbPas):
    if borneInf >= borneSup:
        print("Erreur : La borne inférieure doit être strictement inférieure à la borne supérieure.")
        return {} 
    
    resultats = {}
    x = borneInf

    print(f"Évaluation de la fonction sur [{borneInf}, {borneSup}] avec un pas de {nbPas}")
    print(f"x | f(x)")
    print("-----------")

    while x <= borneSup:
        y = calc_x(x)
        print(f"{x} | {y}")
        resultats[round(x, 2)] = y 
        x += nbPas

    return resultats

mon_dictionnaire = poly(calc_x, borneInf=-10, borneSup=20, nbPas=1)
print(f"Voici les différentes valeurs enregistrées dans mon dictionnaire {mon_dictionnaire}.")
print(f"Si x est égal à {mon_dictionnaire.keys}) alors f(x) sera égal à {mon_dictionnaire.values}.")
    
```

```python
# Écrire une fonction compterMots ayant un argument (une chaîne de caractères) et qui renvoie un dictionnaire qui contient la fréquence de tous les mots de la chaîne entrée.
import re
phrase = input("Veuillez saisir une phrase svp:")
phrase = re.sub(r'[.,;:\'?"()]', ' ', phrase)
phrase = phrase.lower()

def compterMots(phrase):
    dico_mots = {}
    phrase = phrase.split()
    if not phrase:
        print("Veuillez saisir quelque chose svp.")
    else:
        for mot in phrase:
            if mot in dico_mots:
                dico_mots[mot] += 1
            else:
                dico_mots[mot] = 1
    print(dico_mots)   
            
    
compterMots(phrase)
    
    
    
    
    
    
    
```

```python
# Fonction renvoyant plusieurs valeurs sous forme d’un tuple. Écrire une fonction minMaxMoy qui reçoit une liste d’entiers et qui renvoie le minimum, le maximum et la moyenne de cette liste. Le programme principal appellera cette fonc- tion avec la liste : [10, 18, 14, 20, 12, 16].

ma_liste = [10, 18, 14, 20, 12, 16]
def minMaxMoy(liste):
    min_list = min(liste)
    max_list = max(liste)    
    moy_list = sum(liste)/len(liste)
    return  min_list, max_list, moy_list

minMaxMoy(ma_liste)
```
