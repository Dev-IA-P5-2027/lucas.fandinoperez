```python
import re
```

```python
# define a pattern

pattern = r'\d+'

regex_object = re.compile(pattern)
```

```python
text = "there are 100 mangos and 456 bananas."

match = regex_object.search(text)

if match: 
    print("Pattern found:" , match.group())
else: 
    print("Pattern not found")
```

```python
match = re.findall(pattern, text)

print(match)
```

```python
pattern1 = r'orange'

# Le texte dans lequel effectuer la recherche
text = 'I have an orange, apple, and a pineapple'

# Utilisation de re.search pour trouver le motif
result = re.search(pattern1, text)

# Vérification si un résultat a été trouvé
if result:
    print('Yes! I gotcha:', result.group())
else:
    print('Sorry, no pattern is found')
```

```python
# Target String
target_string = "Emma is a badminton player who was born on June 17. Another famous volley ball player is John."

# search() for nine-letter word (as per the code \w{9})
# Note: 'badminton' has 9 letters.
result = re.search(r"\w{9}", target_string)

# Print match object
if result:
    print("Match Object:", result.group())
else:
    print("No match found")
```

```python
# The zip codes in USA have 5 digits.
statement = 'The zip code of Besancon is 25000'

# regex = r'[0-9]{5}' # or we can also use this pattern: '\d{5}'
regex = r'[0-9]{5}'

zip_code = re.findall(regex, statement)

if zip_code:
    print("Zip Code Found! -->", zip_code)
else:
    print('Not Found!')
```

```python
pattern = r'\bhello\b'

# Create a regex object using re.compile()
regex_object = re.compile(pattern)

# Sample text
text = 'Hello, how are you? Saying hello is nice.'

# Using re.search() with the compiled regex object
match = regex_object.search(text)

if match:
    print('Pattern found:', match.group())
else:
    print('Pattern not found')
```

```python
# Target String
target_string = "Emma is a baseball player who was born on June 17"

# findall() for words of 1 to 2 letters
result = re.findall(r"\b\w{1,2}\b", target_string)

# Print the list of matches
print("Matches found:", result)
```

```python
# Define the dictionary of valid words
dictionary = ['am', 'and', 'i', 'you', "you're", 'we', 'are', 'is', 'she', 'happy', 'sad', 'not', 'today', 'but', 'like']

# Input text
text = "We re happy today, but you're not happy lke she is."

# Split the text into words
# match any word with lower case letter(s) or consisting "'"
words = re.findall(r"\b[a-z']+\b", text.lower())

# Iterate through the words and append <typo: word> to those not in the dictionary
for i, word in enumerate(words):
    # Check if the word (lowercase) is not in the dictionary and word.isalpha()
    # Note: word.isalpha() might fail if the word contains an apostrophe like "you're"
    if word.lower() not in dictionary:
        words[i] = f"<typo:{word}>"  # Annotate the typo

# Reconstruct the text
text = ' '.join(words)

print('Result:', text)
```
