```python
# dealing with the contractions
# !pip install contractions

import contractions

def fix_contractions(data):
    expanded_corpus = [contractions.fix(text) for text in data]
    return expanded_corpus
```

```python
corpus = ["I can't wait for the new season of my favourite show !", "Lemme know when you're ready.", "Let's go !", "Nothing gonna change my love for you."]

fix_contractions(corpus)
```

```python
import unicodedata

def stan_accented_chars(text):
    return unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8", "ignore")

text3 = "Some words like résumé, caffè, lattè, à bientôt, divorcée etc..."

stan_accented_chars(text3)
```

```python
# Noise removal 

#1 - removing html tags

def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, "", text)

html_text = "<p>Hello,  <b>world</b>!</p> <a href='#'>Click here</a>"

clean_text = remove_html_tags(html_text)

print(clean_text)
    
```

```python
# removing numbers or digits

def remove_numbers(text):
    result = re.sub(r'\d{7,}', '<masked number>', text)
    return result

text_num= "Her phone number is 01234567 and mine is 0601900219"

remove_numbers(text_num)
```

```python
# convert the numbers into words

!pip install inflect

import inflect

p = inflect.engine()

# convert number into words
def convert_number(text):
    return ' '.join([p.number_to_words(w) if w.isdigit() else w for w in text.split()])

input_text = 'There are 3 balls in this bag, and 12 in the other one.'
input_text2 = " There are about 1000 students who are wnating to enter the school 42"
print(convert_number(input_text2))
```

```python
# removing punctuations and sepcial characters

regular_punct = list(string.punctuation)


def remove_punc(text, punc_list):
    for punc in punc_list:
        if punc in text:
            text = text.replace(punc, '')
    return text.strip

remove_punc("Hey #Lisa! Did you know that the summer break is coming ? Amazing, right ! ! It's only five more days ! ! ", regular_punct)
```

```python
# removal of url 

url_pattern = re.compile(r'https?://\S+')



url_patterns = []
url_patterns.append(re.compile(r'https?://\S+'))
url_patterns.append(re.compile(r'http?://\S+'))
url_patterns.append(re.compile(r'www\.\S+'))

def remove_urls(text):
    for pattern in url_patterns:
        text = pattern.sub('<this link>', text)
    return text


input_text_url = "Visit the following link: www.google.com"
print("URL traitée :", remove_urls(input_text_url))
```

```python
# STOPWORDS REMOVAL

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Téléchargement des ressources nécessaires
nltk.download('stopwords')
nltk.download('punkt')

def remove_stopwords(text, lang='english'):
    # 1. Charger la liste des mots vides pour la langue choisie
    stop_words = set(stopwords.words(lang))
    
    # 2. Tokeniser le texte (le découper en mots)
    word_tokens = word_tokenize(text)
    
    # 3. Filtrer : on ne garde que les mots qui ne sont pas dans la liste
    # (en mettant tout en minuscule pour la comparaison)
    filtered_text = [w for w in word_tokens if w.lower() not in stop_words]
    
    return filtered_text

# TEST
example_text = "This is a sample sentence, showing off the stop words filtration."
print(remove_stopwords(example_text))
# Résultat : ['sample', 'sentence', ',', 'showing', 'stop', 'words', 'filtration', '.']
```

```python
# stemming

import re
import nltk
import inflect
from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize


# --- INITIALISATION ---
p = inflect.engine()
stemmer = PorterStemmer()

# Téléchargement des ressources NLTK nécessaires
nltk.download('punkt')
nltk.download('punkt_tab')
def stem_words(text):
    # Tokenisation : découpe le texte en mots
    word_tokens = word_tokenize(text)
    # Stemming : réduit chaque mot à sa racine (ex: scientific -> scientif)
    stems = [stemmer.stem(word) for word in word_tokens]
    return stems

stem_input = 'data science uses scientific methods algorithms and many types of processes'
print(f"\nStemming result: {stem_words(stem_input)}")
```

```python
def lemmatize_text(text):
    doc = nlp(text)
    # On récupère le lemme de chaque token
    return ' '.join([token.lemma_ for token in doc])

# --- TESTS GLOBAUX ---

input_text = "The quick brown foxes are jumping over 12 lazy dogs at https://google.com"

# Étape A : Nettoyage URL
clean_url = remove_urls(input_text)

# Étape B : Nombres
clean_numbers = convert_number(clean_url)

# Étape C : Lemmatisation (Plus précis que le stemming)
final_lemma = lemmatize_text(clean_numbers)

print(f"Original: {input_text}")
print(f"Final (Clean + Lemma): {final_lemma}")

# Test Stemming séparé (comme dans ton image)
stem_input = 'data science uses scientific methods'
print(f"\nStemming result: {stem_words(stem_input)}")
```
