# Brief projet — Évolution du triple saut (1891 → 2019)

Source : http://trackfield.brinkster.net — événements `MF4` (hommes) / `WF4` (femmes).

On récupère donc le bloc public disponible (« 25 Best Performances » pour les anciens cycles, « Top 10 Athletes » pour les récents).*

```python
# Étape 1 — codes événements + cycles de 4 ans (1891 → 2019)
import time
import pandas as pd

BASE = "http://trackfield.brinkster.net"
EVENTS = {"Hommes": "MF4", "Femmes": "WF4"}
CYCLES = ['1891-1900', '1901-1908', '1909-1912', '1913-1920', '1921-1924', '1925-1928',
          '1929-1932', '1933-1936', '1937-1948', '1949-1952', '1953-1956', '1957-1960',
          '1961-1964', '1965-1968', '1969-1972', '1973-1976', '1977-1980', '1981-1984',
          '1985-1988', '1989-1992', '1993-1996', '1997-2000', '2001-2004', '2005-2008',
          '2009-2012', '2013-2016', '2017-2020']
print(len(CYCLES), "cycles, de", CYCLES[0], "à", CYCLES[-1])
```

```python
# Étape 2 — URL des perfs du cycle contenant 2019 (2017-2020) + fonction de scraping
def url_cycle(idx, periode):
    return f"{BASE}/results.asp?Cycles=29&Cycle={idx}&Years={periode}"

print("URL 2019 :", url_cycle(27, "2017-2020"))

def scraper(event, periode, idx):
    for _ in range(3):  # le serveur coupe parfois la connexion -> on réessaie
        try:
            r = requests.post(
                url_cycle(idx, periode),
                data={"hdnEvent": event, "lstYears": periode, "P": "F", "hdnMedalist": "No"},
                timeout=30,
            )
            break
        except requests.exceptions.RequestException:
            time.sleep(2)
    else:
        return []
    soup = BeautifulSoup(r.content.decode("latin-1"), "html.parser")
    cells = [td.get_text(" ", strip=True) for td in soup.find_all("td")]
    block = ""
    for cle in ["25 Best Performances", "Top 10 Athletes"]:
        trouve = [c for c in cells if cle in c]
        if trouve:
            block = max(trouve, key=len)
            block = block[block.find(cle):].split("World Records")[0]
            break
    return re.findall(r"(\d{1,2})\s+([A-Za-zÀ-ÿ.'\- ]+?)\s+([A-Z]{3})\s+(\d{1,2}\.\d{1,3})", block)[:25]

scraper("MF4", "2017-2020", 27)[:3]
```

```python
# Étape 3 — scraping de tous les cycles (hommes + femmes) → DataFrame
lignes = []
for sexe, code in EVENTS.items():
    for idx, periode in enumerate(CYCLES, start=1):
        for rang, nom, pays, perf in scraper(code, periode, idx):
            lignes.append({"sexe": sexe, "periode": periode, "annee_debut": int(periode[:4]),
                           "rang": int(rang), "athlete": nom.strip(), "pays": pays,
                           "performance_m": float(perf)})
        time.sleep(0.3)

df = pd.DataFrame(lignes)
print(df.shape)
df.head(10)
```

```python
# Étape 4 — export Excel (une feuille par sexe)
fichier = "triple_saut_1891_2019.xlsx"
with pd.ExcelWriter(fichier, engine="openpyxl") as writer:
    for sexe in EVENTS:
        df[df["sexe"] == sexe].to_excel(writer, sheet_name=sexe, index=False)
print("Fichier écrit :", fichier)
```

```python
# Étape 5 — étude statistique
import matplotlib.pyplot as plt

print(df.groupby("sexe")["performance_m"].describe(), "\n")

# Meilleure performance (rang 1) de chaque cycle
best = df[df["rang"] == 1]

fig, ax = plt.subplots(figsize=(11, 5))
for sexe in EVENTS:
    d = best[best["sexe"] == sexe].sort_values("annee_debut")
    ax.plot(d["annee_debut"], d["performance_m"], marker="o", label=sexe)
ax.set_title("Évolution de la meilleure performance en triple saut (1891 → 2019)")
ax.set_xlabel("Année (début de cycle)")
ax.set_ylabel("Performance (m)")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Progression totale par sexe
for sexe in EVENTS:
    d = best[best["sexe"] == sexe].sort_values("annee_debut")
    if len(d):
        print(f"{sexe} : {d['performance_m'].iloc[0]:.2f} m ({d['annee_debut'].iloc[0]}) "
              f"-> {d['performance_m'].iloc[-1]:.2f} m ({d['annee_debut'].iloc[-1]}) "
              f"soit +{d['performance_m'].iloc[-1] - d['performance_m'].iloc[0]:.2f} m")
```
