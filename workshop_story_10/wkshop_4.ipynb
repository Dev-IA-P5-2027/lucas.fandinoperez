# Exercice 1 

```python
# 1. Définir l'URL du fichier robots.txt de Wikipédia
url = "https://fr.wikipedia.org/robots.txt"

# 2. Configurer un User-Agent (Wikipédia bloque souvent les robots anonymes)
headers = {
    "User-Agent": "MonRobotDEtude/1.0 (contact: monadresse@email.com)"
}

# 3. Envoyer la requête GET au serveur
response = requests.get(url, headers=headers)

# 4. Vérifier si la requête a réussi (Statut 200)
if response.status_code == 200:
    print("--- Contenu du fichier robots.txt de Wikipédia ---")
    # On utilise .text car c'est un fichier texte brut
    print(response.text)
else:
    print(f"Échec de la récupération. Code statut : {response.status_code}")
```
