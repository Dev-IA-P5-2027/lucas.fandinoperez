```python
import requests
import re
import html 
from bs4 import BeautifulSoup
import json
```

# PART 1: Using BeautifulSoup

# Get the Data using URL

```python
# Ici, pour nous entraîner, nous utiliserons l'url du site "quotes.toscrape.com", un site spécifiquement conçu pour le scrapping de données avec des citations (pas d'erreur 403, pas de firewall !)

url = "http://quotes.toscrape.com/"


```

```python
#get the page data
page = requests.get(url)
print(f"Nouveau Satut : {page.status_code}.")
print(f"Nouvelle longueur: {len(page.text)}.")
```

```python
page.encoding = "utf-8"
```

```python
#parse the data 

soup = BeautifulSoup(page.text, "html.parser")
print(soup.prettify())
```

```python
citations =  soup.find_all('div', class_= "quote")
print(citations)
```

```python
# REGEX to find a particular class

quote_tag = soup.find_all('span', class_ = re.compile(r'text+'))
print(quote_tag)

author_tag = soup.find_all('small', class_ = re.compile(r'author+'))
print(author_tag)
```

```python
#On crée une liste vide pour stocker tous nos dictionnaires

liste_citations = []

for quote in citations:
    text = quote.find("span", class_= "text").text
    auteur = quote.find("small", class_ = "author").text
    
    dico = {
        "auteur" : auteur,
        "citation" : text,
    }
    
    liste_citations.append(dico)
    
#print(f"Première citation : {liste_citations[0]}.")
print(f"La première citation est: {liste_citations[0]["citation"]} et son auteur est {liste_citations[0]["auteur"]}.")
```

```python
# On peut maintenant convertir toute ta liste en un fichier JSON propre en une seule ligne :

fichier_json = json.dumps(liste_citations, indent=4, ensure_ascii=False)
print("\nVoici ton résultat final structuré :\n", fichier_json) 
```

# PART 2 : USING REGEX ONLY

```python
page.text
quotes = re.findall(r'<span class="text" itemprop="text">(.*?)</span>', page.text)
print(quotes[0])
```

```python
authors = re.findall(r'<small class="author" itemprop="author">(.*?)</small>', page.text)
print(authors)
```

```python
liste_quotes = []

for q, a in zip(quotes, authors):
    
    # On crée UN dictionnaire pour CETTE citation précise
    dico_2 = {
        "citation": q.strip(),
        "auteur": a.strip()
    }
    
    # On l'ajoute à la liste (sans faire "liste = liste.append()")
    liste_quotes.append(dico_2)

print(liste_quotes)


```

```python
# On peut maintenant convertir toute ta liste en un fichier JSON propre en une seule ligne :

fichier_json_2 = json.dumps(liste_quotes, indent=4, ensure_ascii=False)
print("\nVoici ton résultat final structuré :\n", fichier_json) 
```
