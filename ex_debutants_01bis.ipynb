```python
# Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un rayon et la hauter, puis , qui calcule le volume d'un cône droit.

rayon = int(input("Veuillez saisir un rayon en cms svp : "))

def volume_cone(rayon, hauteur):
    volume = 1/3 * 3.141592 * rayon**2 * hauteur
    return volume

print(f"Le volume du cône droit est {volume_cone(rayon, 4)}.")
```

```python
# Écrire dans une cellule un programme qui demande à l'utilisateur son poids et sa taille et qui calcule son IMC (trouver la formule sur internet).

poids = int(input("Veuillez saisir votre poids en kgs svp : "))
taille = float(input("Veuillez saisir votre taille en mètres svp : ").replace(',', '.'))

def calc_imc(poids, taille):
    imc = poids / (taille**2)
    return imc

print(calc_imc(poids, taille))
```

```python
# Écrire une fonction qui convertit en mètres par seconde et en km/h une vitesse fournie en paramètres par l'utilisateur en miles/heure (1 mile = 1609 mètres). Ce résultat sera retourné par la fonction.
miles = int(input("Veuillez saisir une distancevitess en miles/heure svp: "))

def miles_speed (miles):
    kms = miles * 1.60934
    ms = kms / 3.6
    print(f"La vitesse en kms/h est de {kms} et la vitesse en m/s est de {ms}.")

miles_speed(miles)
```

```python
# Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un entier n positif et qui affiche PAIR s'il est divisible par 2, IMPAIR sinon (penser au modulo).
nb_entier = int(input("Veuillez saisir un entier positif n svp : "))

def pair (nb_entier):
    if nb_entier < 0:
        print("Veuillez saisir un nombre positif svp.")
    elif nb_entier % 2 == 0:
        print(f"{nb_entier} est un nombre pair.")
    else:
        print(f"{nb_entier} est un nombre impair.")

pair(nb_entier)
```

```python
# Écrire un programme en python pour réviser ses tables de multiplication. Le programme tire au hasard deux entiers entre 0 et 10 et demande à l'utilisateur de donner leur produit.

import random

def jeu_multiplication():
    print("Bienvenue dans le réviseur de tables de multiplication !")
    print("Tu auras 5 questions. Bonne chance !\n")

    score = 0
    nombre_questions = 5

    for i in range(1, nombre_questions+1):
        nombre1 = random.randint(0, 10)
        nombre2 = random.randint(0, 10)
        reponse_correcte = nombre1 * nombre2
        question = f"Question {i} sur {nombre_questions} : Combien font {nombre1} * {nombre2} ?"
        reponse_utilisateur = int(input(question))
        if reponse_utilisateur == reponse_correcte:
            score += 1
            print("Bonne réponse, vous gagnez un point.")  
        else:
            score -= 1
            print(f"Mauvaise réponse, vous perdez un point. La bonne réponse était {reponse_correcte}")
            
    print(f"Votre score final est de {score} sur 5.")

jeu_multiplication()
  
        
        
```

```python
# L'ordinateur tire un nombre entier au hasard entre 0 et 100.
# L'utilisateur doit le trouver et pour cela propose des valeurs.
# L'ordinateur indique pour chaque valeur proposée si la valeur est trop petite, trop grande ou s'il a trouvé.
# a) Écrire un programme en Python pour jouer à ce jeu.
# b) Modifier le programme pour qu'il s'arrête si l'utilisateur n'a pas trouvé au bout de 10 coups.
# c) Mettre ce programme dans une fonction plus_ou_moins() qui prendra en paramètres :

# le nombre min et max de la plage de tirage du nombre au hasard
# le nombre de coups maximum

import random

nombre_questions = 10
ecart_valeurs = range(0, 101)

def plus_ou_moins(nombre_questions, ecart_valeurs):
    print("Bienvenue dans le jeu, tu dois deviner un nombre au hasard entre 0 et 100 !\n")
    print("Tu auras 10 tentatives pour le deviner, à chaque fois l'ordinateur te diras si le nombre est plus grand ou plus petit.\n")
    print("Bonne chance !\n")
    nombre_mystere = random.randint(0, 100)
    

    for i in range(1, nombre_questions+1):
        reponse_correcte = nombre_mystere
        question = f"Question {i} sur {nombre_questions} : Quel est le nombre mystère ?"
        reponse_utilisateur = int(input(question))
        if reponse_correcte == reponse_utilisateur:
            print("Bravo vous avez deviné, vous gagnez la partie.")
            break
        elif reponse_utilisateur < reponse_correcte:
            print(f"Vous proposez: {reponse_utilisateur}. Le nombre mystère est plus grand, cherchez encore.")
        elif reponse_utilisateur > reponse_correcte:
            print(f"Vous proposez: {reponse_utilisateur}. Le nombre mystère est plus petit, cherchez encore.")
        if i == 10:
            print(f"GAME OVER. Le nombre mystère était {reponse_correcte}.")
            break
        

plus_ou_moins(nombre_questions, ecart_valeurs)
```

```python
# Ecrire un programme qui tourne indéfiniement et qui demande à l'utilisateur de rentrer un nombre entier compris dans une plage définie aléatoirement (par exemple, entrer un nombre compris entre 12 et 34).
# Si le nombre entré est en dehors de la plage, le programme l'affiche et redemande à nouveau un nombre dans la même plage.
# Si le nombre entré est dans la plage, on félicite l'utilisateur et le programme reprend avec un nouvelle plage.

import random


def jeu_plage_nb():
    print("Bienvenue dans ce jeu ! \n")
    print("A chaque tour, tu dois entrer un nombre compris dans une plage définie aléatoirement jusqu'à trouver la bonne réponse. \n")
    print("Si ton nombre est dans la page définie, une nouvelle plage sera générée au tour d'après. \n")
    
    while True:
        borne_a = random.randint(1,201)
        borne_b = random.randint(1,201)
        val_min = min(borne_a, borne_b)
        val_max = max(borne_a, borne_b)

        print(f"\n--- Nouveau Tour ---")

        while True:
            reponse_utilisateur = int(input("Entrez un nombre entier entre 1 et 200:"))
            if val_min <= reponse_utilisateur <= val_max :
                print(f"Bien joué, {reponse_utilisateur} est bien entre {val_min} et {val_max}. Génération d'une nouvelle plage de nombres.")
                break
            else:
                print(f"Le nombre {reponse_utilisateur} n'est pas dans la plage définie. Essayez encore")
                
            

jeu_plage_nb()
```

```python
# Écrire un programme permettant d'entrer des notes d’élèves sur 20 jusqu’à ce l’utilisateur saisisse une note vide. Cette fonction retourne le nombre de notes entrées, la note la plus élevée, la note la plus basse, la moyenne de toutes les notes.

def note_sur_20():
    liste_notes = []
    print("Saisissez les notes sur 20 svp. Appuyez sur ENTRÉE sans rien écrire pour terminer.")
    
    while True:
        saisie = input("Veuillez saisir une note sur 20 : ")
        if not saisie.strip():
            print("Fin de la saisie des notes.")
            break
        saisie = saisie.replace(",", ".")
        try:
            note = float(saisie)
        except ValueError:
            print("Erreur : Veuillez saisir un nombre valide.")
            continue  
        if not (0 <= note <= 20):
            print("⚠️ Attention : La note doit être comprise entre 0 et 20.")
            continue  # Revenir au début de la boucle
        liste_notes.append(note)
        print("Note ajoutée.")
    if liste_notes:   
        print("--------------------------------------------------------")
        print(f"Le nombre total de notes s'élève à {len(liste_notes)}.")
        print(f"La note maximale est de {max(liste_notes)} sur 20.")
        print(f"La note minimale est de {min(liste_notes)} sur 20.")
        print(f"La moyenne du groupe est de {sum(liste_notes)/len(liste_notes)}sur 20.")
    else:
        print("Aucune note valide n'a été entrée. Fin de la saisie des notes.")
            


note_sur_20()    
```
