*** Placer des pixels sur une image**

```python
#Créer une image de 800x600 pixels, avec un fond gris clair (lightgrey).
# 2.1. placer des pixels noirs aux coordonnées (200,100), (500,200), (600,500) et (700,400)

image_2 = Image.new("RGB", (800, 600), "lightgrey")
pixels_a_modifier = [
    (200, 100),
    (500, 200),
    (600, 500),
    (700, 400)
]

#couleur
couleur_noire = (0, 0, 0)

#boucle
for x, y in pixels_a_modifier:
    image_2.putpixel((x, y), couleur_noire)

image_2.show()
```

```python
# 2.2. tracer une ligne noire horizontale, de la largeur de l’image, à la coordonnée y=250

pixels = image_2.load()

draw = ImageDraw.Draw(image_2)

y_position = 250

draw.line(
    [(0, y_position), (800 - 1, y_position)],
    fill = couleur_noire, 
    width = 2
)

```

```python
# 2.3. tracer une ligne noire verticale, de la hauteur de l’image, à la coordonnée x=250

x_position = 250

draw.line(
    [(x_position, 0), (x_position, 600 -1)], 
    fill = couleur_noire,
    width = 2
)

```

```python
# 2.4. tracer un canevas de lignes bleues, horizontales, distantes de 60 pixels

couleur_bleue = (0, 0, 255) 
espacement = 60
epaisseur_ligne = 1


for y in range(0, 600, espacement):
    draw.line(
        [(0, y), (800 - 1, y)],
        fill=couleur_bleue,
        width=epaisseur_ligne
    )
image_2.show()
```

```python
# 2.5. y ajouter un canevas de lignes verticales vertes, distantes de 60 pixels, pour former un 
# quadrillage

couleur_verte = (0, 255, 0) 

for x in range(0, 800, espacement):
    draw.line(
        [(x, 0), (x, 600-1)],
        fill=couleur_verte,
        width=epaisseur_ligne
    )

image_2.show()
```

```python
# 2.6. tracer un carré jaune, de 200 pixels de côté, aux coordonnées (500,100)
couleur_jaune = (255, 255, 0) 
taille_cote = 200



draw.rectangle(
[(500, 100), (700, 300)],
fill= couleur_jaune, 
width= epaisseur_ligne
)


image_2.show()
```

```python
# 2.7 . remplir l’image de carrés violets, de 30 pixels de côté, distants de 30 pixels les uns des 
# autres, en commençant aux coordonnées (10,10)

couleur_violette = (255, 0, 255) 
taille_cote_2 = 30
distance = 30

for y in range(10, 600, 60):
    for x in range(10, 800, 60):
        x1 = x
        y1 = y
        
        x2 = x + taille_cote_2
        y2 = y + taille_cote_2
        draw.rectangle(
            [(x1, y1), (x2, y2)], 
            fill = couleur_violette, 
            width = epaisseur_ligne
        )

image_2.show()
```

*** 3. FABRICATION DE DRAPEAUX ***

```python
#  Dessiner le drapeau français.
 
drapeau = Image.new("RGB",(600,400),"white")
draw_drapeau = ImageDraw.Draw(drapeau)

bleu_france = (0, 38, 117)
rouge = (239, 65, 53)


draw_drapeau.rectangle(
    [(0, 0), (200, 400)], 
    fill = bleu_france,
)

draw_drapeau.rectangle(
    [(400, 0), (600, 400)],
    fill=rouge
)

drapeau.show()
```

```python
# .2. Dessiner le drapeau de la Belgique.

drapeau_belge = Image.new("RGB",(600,400),"yellow")
draw_drapeau_b = ImageDraw.Draw(drapeau_belge)

draw_drapeau_b.rectangle(
[(0, 0), (200, 400)], 
    fill = couleur_noire,
)

draw_drapeau_b.rectangle(
[(400, 0), (600, 400)], 
    fill = rouge,
)

drapeau_belge.show()
```
