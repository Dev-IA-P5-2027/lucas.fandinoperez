# Détection des bords

* Sobel
* Canny

```python
# utiliser la methode sobel pour détecter les bords
# Vous pouvez utiliser d'autres images de votre choix

# 1. Gaussian blur pour enlever "grain" léger de l'image (meilleurs résultats par la suite)

sudoku_smooth = cv2.GaussianBlur(sudoku, (3, 3), 0)

# 2. Calculer le gradient selon X et Y
# cv2.CV_64F est utilisé pour éviter de perdre les gradients négatifs
sobelx = cv2.Sobel(sudoku_smooth, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(sudoku_smooth, cv2.CV_64F, 0, 1, ksize=3)

# 2. Combiner les deux (Magnitude)
sobel_combined = cv2.magnitude(sobelx, sobely)

# 3. Convertir en format affichable (uint8)
sobel_final = np.uint8(sobel_combined)


cv2.imshow('Sobel Detect', sobel_final)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
# Canny

# Application de Canny
# on réutilise sudoku_smooth (blur Gaussien)
#seuil bas: 30 / seuil haut: 100 puis respectivement 15/75 (meilleure combinaison)


edges = cv2.Canny(sudoku_smooth, 30, 100)

cv2.imshow('Canny Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()


```

```python
# charger l'image 02
# faites une recherche sur internet pour détecter les contours des objets dans image_02

img_couleur = cv2.imread("image_02.png")
#on acharge en gris
img_02 = cv2.imread("image_02.png", 0)

cv2.imshow("image2", img_02)
cv2.waitKey(3000)
cv2.destroyAllWindows()

#on applique un flou Gaussien

img02_smooth = cv2.GaussianBlur(img_02, (3, 3), 0)

# on applique un Canny (binarisation)

edges = cv2.Canny(img02_smooth, 30, 100)

#on utilise edges pour détecter les contours depuis l'image binarisée

contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# On dessine sur 'img_02' pour voir les contours en noir (0, 0, 0)
cv2.drawContours(img_couleur, contours, -1, (0, 0, 0), 2)

# Affichage
cv2.imshow("Contours detectes", img_couleur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#on charge en gris
img3 = cv2.imread("image_03.jpg", 0)

cv2.imshow("image", img3)
cv2.waitKey(3000)
cv2.destroyAllWindows()

#on applique un flou Gaussien

img_smooth = cv2.GaussianBlur(img3, (3, 3), 0)

# on applique un Canny (binarisation)

edges = cv2.Canny(img_smooth, 30, 100)

#on utilise edges pour détecter les contours depuis l'image binarisée

contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# On transforme l'image grise en image "fausse couleur" 
# pour qu'elle puisse accepter le vert (0, 255, 0)

img_color = cv2.cvtColor(img3, cv2.COLOR_GRAY2BGR)

# On dessine sur l'image de base copiée pour voir les contours en vert 
cv2.drawContours(img_color, contours, -1, (0, 255, 0), 2)

# Affichage
cv2.imshow("Contours detectes", img_color)
cv2.waitKey(3000)
cv2.destroyAllWindows()
```

```python
import random

# PROBLEME: UN SEUL OBJET EST DETECTE ET NON 7 (les frontières entre les différents objets touchants disparaissent)

# On utilise l'algorithme watershed (ligne de partage des eaux) pour avoir 7 objets circulaires et non pas une seule forme abstraite

############ PRE-TRAITEMENT ####################

#OTSU
_, thresh = cv2.threshold(img3, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Ouverture morphologique pour éliminer les bruits résiduels
kernel = np.ones((3,3), np.uint8)
morph = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=2)

# --- Étape A : Identifier le FOND sûr (sure background) ---

sure_bg = cv2.dilate(morph, kernel, iterations=3)

# --- Étape B : Identifier l'OBJET sûr (le centre) ---
# c'est ce qu'on appelle la Transformée de Distance

dist_transform = cv2.distanceTransform(morph, cv2.DIST_L2, 5)

# On seuille la distance pour ne garder que le "cœur" de chaque forme
# Plus le seuil est haut (ici 0.5 * max), plus les cœurs sont petits

_, sure_fg = cv2.threshold(dist_transform, 0.5 * dist_transform.max(), 255, 0)
sure_fg = np.uint8(sure_fg)

# --- Étape C : Identifier la zone INCONNUE ---
# C'est la zone où se situent les contours qui se touchent

inconnue = cv2.subtract(sure_bg, sure_fg)

# --- Étape D : Création des MARQUEURS ---
# On labellise les centres des objets (0 pour le fond, 1, 2, 3... pour les objets)

_, markers = cv2.connectedComponents(sure_fg)

# On ajoute 1 à tous les labels pour que le fond sûr ne soit pas 0 mais 1
# Et les objets soient 2, 3, 4...
markers = markers + 1

# Maintenant, on marque la zone INCONNUE avec 0
markers[inconnue == 255] = 0


# --- Étape E : Application de l'algorithme WATERSHED ---
# L'algorithme a besoin de l'image originale COULEUR et des marqueurs

img_color = cv2.cvtColor(img3, cv2.COLOR_GRAY2BGR)
markers = cv2.watershed(img_color, markers)

# --- VISUALISATION DU RÉSULTAT ---
# Les contours des séparations sont marquées par -1 dans 'markers'
# On les dessine en bleu sur l'image couleur
img_color[markers == -1] = [255, 0, 0]

# Pour mieux voir, colorons chaque objet avec une couleur aléatoire
# On crée une palette de 256 couleurs aléatoires
colors = []
for i in range(256):
    colors.append((random.randint(0, 255), random.randint(0, 255), random.randint(0, 255)))

# On applique les couleurs aléatoires aux marqueurs
color_segmentation1 = np.zeros_like(img_color)
for m in range(2, markers.max() + 1):
    color_segmentation1[markers == m] = colors[m]


cv2.imshow("Segmentation coloree", color_segmentation1)
cv2.waitKey(3000)
cv2.destroyAllWindows()
cv2.imwrite("image03_random_color_edged.png", color_segmentation1)


```

```python



# 1. Le filtre Médian "Musclé"
# Un noyau de 3 est trop faible. On passe à 5 pour forcer 
# le remplacement des groupes de pixels bruités.
median = cv2.medianBlur(mona_org, 5)

# 2. L'astuce secrète : Ouverture Morphologique
# L'ouverture élimine les petits éléments blancs (sel) sur fond noir
# La fermeture élimine les petits trous noirs (poivre)
kernel = np.ones((3,3), np.uint8)
cleaned = cv2.morphologyEx(median, cv2.MORPH_OPEN, kernel)
cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_CLOSE, kernel)

# 4. Finition NL-Means (maintenant qu'il y a moins de gros blocs)
final = cv2.fastNlMeansDenoisingColored(cleaned, None, 15, 7, 21)

cv2.imshow("Mona Enfin Propre", final)
cv2.waitKey(3000)
cv2.destroyAllWindows()
cv2.imwrite("meilleure_version.png", final)
```

```python
# 1. Créer une version très floue pour capturer les basses fréquences
gauss = cv2.GaussianBlur(final, (5, 5), 0)

# 2. L'Unsharp Masking : amplifier les hautes fréquences (les bords)
# La formule : Résultat = 1.5 * Original - 0.5 * Flou
# 'mona_clean' est ton image actuelle, pas l'originale bruitée
mona_sharp = cv2.addWeighted(final, 1.5, gauss, -0.5, 0)

mona_galerie = cv2.bilateralFilter(mona_sharp, 9, 75, 75)

# --- Affichage Final ---

cv2.imshow("Restauration Galerie d'Art", mona_galerie)
cv2.waitKey(3000)
cv2.destroyAllWindows()
cv2.imwrite("version_galerie.png", mona_galerie)
```

```python
# 1. On repart de ta "meilleure_version" (celle qui est lisse et propre)
mona_lisse = cv2.imread("meilleure_version.png")

# 2. On crée le masque des bords (Edges)
# On veut que la netteté ne s'applique QUE sur les traits importants
edges = cv2.Canny(mona_lisse, 50, 150)
edges_inv = cv2.bitwise_not(edges) # Pour avoir les bords en noir

# 3. Création de l'image nette (Unsharp Mask)
gauss = cv2.GaussianBlur(mona_lisse, (5, 5), 0)
mona_sharp = cv2.addWeighted(mona_lisse, 2.0, gauss, -1.0, 0)

# 4. LE SECRET : On ne garde la version nette QUE là où il y a des bords
# Le reste (le fond, la peau lisse) garde la version propre sans bruit
final = np.where(edges[:,:,None] > 0, mona_sharp, mona_lisse)

# 5. Un dernier petit Bilateral pour lier le tout
final_galerie = cv2.bilateralFilter(final, 5, 50, 50)

cv2.imshow("Mona Lisa Masterpiece", final_galerie)
cv2.waitKey(3000)
cv2.destroyAllWindows()
```

```python
from PIL import Image, ImageFilter, ImageEnhance
 
img = Image.open("mona.png").convert("L")
 
thresholds = [(240, 15), (235, 20), (230, 25)]
 
for white_thresh, black_thresh in thresholds:
    pixels = img.load()
    w, h = img.size
    for y in range(1, h - 1):
        for x in range(1, w - 1):
            p = pixels[x, y]
            if p > white_thresh or p < black_thresh:
                neighbors = []
                for dy in [-1, 0, 1]:
                    for dx in [-1, 0, 1]:
                        if dx == 0 and dy == 0:
                            continue
                        n = pixels[x + dx, y + dy]
                        if black_thresh <= n <= white_thresh:
                            neighbors.append(n)
                if neighbors:
                    pixels[x, y] = sum(neighbors) // len(neighbors)
 
img = img.filter(ImageFilter.MedianFilter(3))
img = img.filter(ImageFilter.UnsharpMask(radius=2, percent=120, threshold=3))
img = ImageEnhance.Contrast(img).enhance(1.3)
 
plt.imshow(img, cmap='gray')    # pour enlever la couleur verte avec matplotlib
plt.axis('off')
plt.show()
```
